```python
#printing the elements in a set
set={"abd","fjf","hvfh","fhbf","fhf"}
print(set)
```

```python
#union of 2 sets
t = {23,53,22,456,43,76,43,5,53,34}
y={32,44,5,33,88,7,65,765}
c=t.union(y)
print(c)
```

```python
#intersection of 2 sets
t = {23,53,22,456,43,76,43,5,53,34}
y={32,44,5,33,88,7,65,765}
c=t.intersection(y)
print(c)
```

```python
#difference between set
t = {23,53,22,456,43,76,43,5,53,34}
y={32,44,5,33,88,7,65,765,76}
c=t-y
print(c)
```

```python
#symmetric difference of 2 sets s
t = {23,53,22,456,43,76,43,5,53,34}
y={23,53,22,43,244,35,32}
print(t.symmetric_difference(y))
```

```python
#checking if a set is a subset of another
t = {23,53,22,456,43,76,43,5,53,34}
y={23,53,22}
print(y.issubset(t))
```

```python
#adding an element to a set
t = {23,53,22,456,43,76,43,5,53,34}
t.add(2)
print(t)
```

```python
#printing the elements of a set as a list
t = {23,53,22,456,43,76,43,5,5,34}
for x in t:
  print(x)
```
